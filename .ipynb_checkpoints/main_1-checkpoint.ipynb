```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv("tmdb_5000_movies_1.csv")
credits = pd.read_csv("tmdb_5000_credits.csv", usecols=['movie_id', 'title', 'cast', 'crew'],low_memory=False)
```

```python
movies=movies.merge(credits, on = "title")
```

```python
movies.shape
```

```python
# genres
# title
# id
# overview
# cast
# crew
movies=movies[['genres','title','movie_id','overview','keywords','cast','crew']]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
def convert(obj):
    l = []
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
movies['genres']=movies.genres.apply(convert)
```

```python
movies['keywords']=movies.keywords.apply(convert)
```

```python
movies.head(1)
```

```python
def convert3(obj):
    l = []
    counter=0
    for i in ast.literal_eval(obj):
        if counter != 3:
            l.append(i['name'])
            counter+=1
        else:
            break
    return l
```

```python
movies['cast']=movies.cast.apply(convert3)
```

```python
movies['cast'].head
```

```python
import ast

def fetch_director(obj):
    L = []
    # Agar data khali (NaN/Null) hai ya string nahi hai, toh khali list bhej do
    if not isinstance(obj, str):
        return L
        
    try:
        for i in ast.literal_eval(obj):
            if i['job'] == 'Director':
                L.append(i['name'])
                break  # Pehla director milte hi ruk jao
    except (ValueError, SyntaxError):
        # Agar parsing mein koi dikkat aaye toh handle ho jaye
        pass
        
    return L
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x: x.split())
```

```python
movies.head()
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
new_df = movies[['movie_id','title','tags']]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.head()
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y) 
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```
